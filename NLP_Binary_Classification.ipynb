## **Logistic Regression + CVEC + BOW**

```python
def text_process(text):

    # Word Tokenize
    final = word_tokenize(text)

    # # Split word with Space
    final = " ".join(word for word in final)

    return final
```

```python
df['text_token'] = df['question'].apply(text_process)
df
```

```python
df['text_token'][0]
```

```python
# Split Train and Test
from sklearn.model_selection import train_test_split
```

```python
X = df[['text_token']]
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.2,
                                                    random_state=25)
```

```python
# Use Count Vectorizer
from sklearn.feature_extraction.text import CountVectorizer

cvec = CountVectorizer(analyzer=lambda x:x.split(' '))
cvec.fit_transform(X_train['text_token'])

cvec.vocabulary_
```

```python
train_bow
```

```python
# Create Bag-Of-Words (BoW)
train_bow = cvec.transform(X_train['text_token'])

pd.DataFrame(train_bow.toarray(), columns=cvec.get_feature_names_out(),
             index=X_train['text_token'])
```

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(train_bow, y_train)
```

```python
from sklearn.metrics import classification_report

test_bow = cvec.transform(X_test['text_token'])
test_prediction = lr.predict(test_bow)

print(classification_report(test_prediction, y_test))
```

### **Error Analysis**

```python
def print_fPred(input = X_test['text_token'], prediction = test_prediction, label = y_test):
  false_pred = pd.DataFrame(columns=['INPUT', 'PREDICTION', 'LABEL'])
  for in_put, predic, la_ble in zip(input, prediction, label):
    if predic != la_ble:
      false_pred = false_pred.append({'INPUT':in_put, 'PREDICTION':predic, 'LABEL':la_ble}, ignore_index = True)
  return false_pred
```
